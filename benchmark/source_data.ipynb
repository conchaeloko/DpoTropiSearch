#### Count prophages per KL type

```python
import pandas as pd
import os 
from collections import Counter

path_repo = "/media/concha-eloko/Linux/PPT_clean"
path_source_file = "/home/concha-eloko/Documents/Projects/Thesis/PPT_R4_submission/source_data"

data_df = pd.read_csv(os.path.join(path_repo, "TropiGATv2.final_df_v2.tsv"), sep = "\t", header = 0)
data_df_clean = data_df[~data_df["KL_type_LCA"].str.contains("\\|")]
df_prophages = data_df_clean.drop_duplicates(subset=["Phage"], keep="first")
```

```python
dico_count_kltypes = dict(Counter(df_prophages["KL_type_LCA"]))
count_KLtypes_df = pd.DataFrame.from_dict(dico_count_kltypes, orient="index", columns=["count"])
count_KLtypes_df.to_csv(f"{path_source_file}/count_KLtypes.tsv", sep="\t")
```
